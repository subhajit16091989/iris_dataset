# 01 — Exploratory Data Analysis (EDA)

**Goal:** Understand the dataset *before* touching any model.

EDA answers four questions:
1. What does the data look like? (shape, types, samples)
2. Is the target balanced?
3. Are there missing values or data quality issues?
4. Which features look informative?

> **Rule:** No model code here. Findings drive decisions in `02_feature_experiments.ipynb` and ultimately in `src/features/preprocess.py`.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def load_data(path: str) -> pd.DataFrame:
    df=pd.read_csv(path)
    return(df)

def validate_data(df: pd.DataFrame):
    if df.empty:
        raise ValueError("Dataset is Empty")  # raise ValueError “Stop the program and throw an error because the input value is invalid”
    missing = df.isnull().sum().sum() ## 
    if missing > 0:
        print(f"Warning: {missing} missing values found")

def generate_summary(df: pd.DataFrame, output_path: str):
    with open(output_path,"w") as f:
        f.write("Shape:\n")
        f.write(str(df.shape) + "\n\n")

        f.write("Data Types:\n") 
        f.write(str(df.dtypes) + "\n\n")

        f.write("Describe:\n") 
        f.write(str(df.describe()) + "\n\n")

## Target Distribution
def plot_target_distribution (df: pd.DataFrame, output_dir: str):
    df["target_name"].value_counts().plot(kind="bar")
    plt.title("Target Distribution")
    plt.savefig(f"{output_dir}/target_distribution.png")
    plt.close

# Feature Distribution
def plt_feature_distribution (df: pd.DataFrame, output_dir: str):
    os.makedirs(output_dir, exist_ok=True)
    numerical_cols = df.select_dtypes(include="number").columns.drop("target")
    for col in numerical_cols:
        plt.figure
        #plot Histogram
        df[col].hist(bins=20, color='steelblue', edgecolor='white')
        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")

        # Clean filename (important for OS compatibility)
        safe_col = col.replace(" ", "_").replace("(", "").replace(")", "")
        file_path = os.path.join(output_dir, f"{safe_col}_distribution.png")

        plt.savefig(file_path)
        plt.close()

    print(f"Feature distribution plots saved to: {output_dir}")

# CoRelation Analysis
def corlation_analysis (df: pd.DataFrame):
    corr_matrix = df.drop(columns=["target", "target_name"]).corr()
    print("\nCorrelation Matrix:\n")
    print(corr_matrix)
    return corr_matrix


def main():
    input_dataset = '../data/raw/iris_dataset.csv'
    output_dir    = '../reports/EDA'
    summary_path = "../reports/EDA/eda_summary.txt"
    df = load_data(input_dataset)
    validate_data(df)

    generate_summary(df, summary_path)
    plot_target_distribution(df,output_dir)
    plt_feature_distribution(df,output_dir)
    corlation_analysis(df)

if __name__ == "__main__":
    main()
```
